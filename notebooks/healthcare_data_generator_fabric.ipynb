# Healthcare Data Generator — Fabric Notebook

This notebook runs the existing generator in `src/` inside Microsoft Fabric.

```python
import os, sys
sys.path.insert(0, 'src')
print('Working directory:', os.getcwd())
try:
    from config_local import CONNECTION_STRING as LOCAL_CONNECTION_STRING
    os.environ['CONNECTION_STRING'] = LOCAL_CONNECTION_STRING
    print('Set CONNECTION_STRING from src/config_local.py')
except Exception:
    print('No CONNECTION_STRING from config_local.py')
```

```python
import importlib
required = ['pandas','sqlalchemy','pyodbc','faker']
missing = [p for p in required if importlib.util.find_spec(p) is None]
if missing:
    print('Missing packages:', missing)
else:
    print('All required packages present')
    import pyodbc
    print('ODBC Drivers:', [d for d in pyodbc.drivers()])
```

```python
import subprocess, sys
args = [sys.executable, 'src/main.py', '--rebuild-start', '2025-01-01']
print('Running:', ' '.join(args))
proc = subprocess.run(args)
print('Exit code:', proc.returncode)
```

# Healthcare Data Generator — Fabric Notebook

This notebook wraps the existing generator in `src/` so you can run it inside Microsoft Fabric. It:
